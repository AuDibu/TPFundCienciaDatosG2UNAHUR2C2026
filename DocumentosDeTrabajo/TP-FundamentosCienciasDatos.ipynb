#### Limpieza inicial del dataset

*Puntos a atacar:*

- Repetidos:
  - Detectados: 
    - 50 filas repetidas
  - Estrategia:
    - Se dejara una fila única por cada repetido
- Conflictos en columnas categóricas:
  - Detectados:
    - Industry presenta inconsistencias en su uso de mayúsculas y minúsculas
    - Salary_before_usd y salary_after_usd: floats que estan en formato string y presentan uso inconsistente del carácter "$"
  - Estrategia:
    - Industry sera pasado a solo minúscula de manera general
    - Salary_before_usd y salary_after_usd seran convertidos a float
- Vacios:
  - Detectados:
    - salary_after_usd: 754
    - salary_before_usd: 753
    - automation_risk_percent: 752
    - industry: 751
  - Estrategia:
    - Los casos vacios de Industry se completaran con "unknown"
    - Los vacios de automation_risk_percent seran completados analizando el promedio a nivel job_role/industry.
    - Los vacios de salary_before_usd y salary_after_usd seran completados analizando la mediana a nivel job_role/industry.
      - Se usa la mediana ya que los sueldos muy altos o muy bajos harian del promedio un valor poco representativo. 

```python

# Limpieza de duplicados
df = df.drop_duplicates()

muestra_df_pre_conversion = df.isnull().sum().copy()

# Saneamiento de strings
df['industry'] = df['industry'].str.lower().str.strip()

for col in ['salary_before_usd', 'salary_after_usd']:
    df[col] = df[col].str.replace(r'[^\d.]+', '', regex=True).astype(str)

df['salary_before_usd'] = pd.to_numeric(df['salary_before_usd'], errors='coerce')
df['salary_after_usd'] = pd.to_numeric(df['salary_after_usd'], errors='coerce')

muestra_df_post_conversion = df.isnull().sum().copy()

if (muestra_df_pre_conversion.equals(muestra_df_post_conversion)):
    print("Se mantiene la cantidad de nulos pre y post conversión.")
else:
    print("La conversión genero nulos.")
    
```

```python

df['industry'] = df['industry'].fillna('unknown')

# --- 2. automation_risk_percent: usar promedio por job_role + industry ---
# Calcular el promedio por grupo (job_role, industry)
agg_stats = df.groupby(['job_role', 'industry'])['automation_risk_percent'].mean().reset_index()
agg_stats.columns = ['job_role', 'industry', 'avg_automation_risk']

# Filtar solo donde hay nulos en la columna objetivo
mask_null = df['automation_risk_percent'].isnull()

# Imputar con el promedio del grupo correspondiente
df.loc[mask_null & (df['job_role'].notna()) & (df['industry'].notna()), 'automation_risk_percent'] = \
    df.loc[mask_null & (df['job_role'].notna()) & (df['industry'].notna()), ['job_role', 'industry']].merge(agg_stats, on=['job_role', 'industry'])['avg_automation_risk'].values

# Verificar si quedan nulos (por columnas sin grupo correspondiente)
print(f"  automation_risk_percent: {df['automation_risk_percent'].isnull().sum()} nulos restantes")

# --- 3. salary_before_usd y salary_after_usd: usar mediana por job_role + industry ---
cols_salary = ['salary_before_usd', 'salary_after_usd']

for col in cols_salary:
    # Calcular la mediana por grupo (job_role, industry)
    agg_stats = df.groupby(['job_role', 'industry'])[col].median().reset_index()
    agg_stats.columns = ['job_role', 'industry', 'mediana_' + col]
    
    # Imputar con la mediana del grupo correspondiente
    mask_null = df[col].isnull()
    df.loc[mask_null & (df['job_role'].notna()) & (df['industry'].notna()), col] = \
        df.loc[mask_null & (df['job_role'].notna()) & (df['industry'].notna()), ['job_role', 'industry']].merge(agg_stats, on=['job_role', 'industry'])['mediana_' + col].values
    
    # Rellenar con 0 los que aún tengan nulos (sin grupo correspondiente)
    df[col] = df[col].fillna(0)

print(f"  salary_before_usd: {df['salary_before_usd'].isnull().sum()} nulos restantes")
print(f"  salary_after_usd: {df['salary_after_usd'].isnull().sum()} nulos restantes")

# --- Resumen de imputación ---
print("\n✅ IMPUTACIÓN COMPLETA:")
for col in ['industry', 'automation_risk_percent'] + cols_salary:
    print(f"  - {col}: {df[col].isnull().sum()} nulos restantes")

# Validación: mostrar filas que fueron imputadas
print("\n🔍 Filas con imputaciones realizadas (primeras 10):")
imputed_rows = df[df['industry'] == 'unknown'].head(10)
display(imputed_rows[['job_role', 'industry', 'salary_before_usd', 'salary_after_usd', 'automation_risk_percent']].head())
```
